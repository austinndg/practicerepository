```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("clean_country_summary.csv")

fig = px.choropleth(
    df,
    locations="country",
    locationmode="country names",
    color="osm_facilities",
    hover_name="country",
    hover_data={
        "country": False,
        "osm_facilities": True,
        "total_footprint_m2": True,
        "continent": True
    },
    color_continuous_scale="Viridis",
    title="Global Data Center Facilities by Country"
)


fig.update_layout(
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection_type="equirectangular"
    ),
    margin={"r":0,"t":40,"l":0,"b":0}
)


fig.show()
```
